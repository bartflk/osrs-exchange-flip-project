```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

root = Path.cwd()
data_dir = root / "data"
if not data_dir.exists():
    data_dir = root.parent / "data"

csv_files = {
    "chaos_rune_562_1year_daily": data_dir / "chaos_rune_562_1year_daily.csv",
    "chaos_rune_562_lastweek_minute": data_dir / "chaos_rune_562_lastweek_minute.csv",
    "item_562_last_24_hours": data_dir / "item_562_last_24_hours.csv",
}

frames = {}
for name, path in csv_files.items():
    frames[name] = pd.read_csv(path)

frames
```

```python
import pandas as pd
import numpy as np

# ---------------------------------------------------------
# LOAD DATA CORRECTLY
# ---------------------------------------------------------
df = frames["chaos_rune_562_lastweek_minute"].copy()

# ds is already ISO8601 with timezone → convert safely
df["ds"] = pd.to_datetime(df["ds"], errors="coerce")

# Basic cleanup
df = df.sort_values("ds")
df = df.reset_index(drop=True)

# ---------------------------------------------------------
# COLUMN SHORTCUTS
# ---------------------------------------------------------
price = "avgHighPrice"
sell = "avgLowPrice"
margin = "margin"
volume = "daily_volume"
profit = "potential_profit"
mxv = "margin_x_volume"

# ---------------------------------------------------------
# PRICE FEATURES
# ---------------------------------------------------------
df["price_change"] = df[price].diff()
df["price_pct_change"] = df[price].pct_change()

df["rolling_mean_5"] = df[price].rolling(5).mean()
df["rolling_mean_30"] = df[price].rolling(30).mean()
df["rolling_std_30"] = df[price].rolling(30).std()

df["zscore_price"] = (df[price] - df["rolling_mean_30"]) / df["rolling_std_30"]

# ---------------------------------------------------------
# MARGIN FEATURES
# ---------------------------------------------------------
df["margin_change"] = df[margin].diff()
df["margin_pct_change"] = df[margin].pct_change()

df["rolling_margin_mean"] = df[margin].rolling(30).mean()
df["rolling_margin_std"] = df[margin].rolling(30).std()

# ---------------------------------------------------------
# VOLUME FEATURES
# ---------------------------------------------------------
df["volume_change"] = df[volume].diff()
df["volume_pct_change"] = df[volume].pct_change()

df["rolling_volume_mean"] = df[volume].rolling(30).mean()
df["rolling_volume_std"] = df[volume].rolling(30).std()

# ---------------------------------------------------------
# PROFIT FEATURES
# ---------------------------------------------------------
df["profit_change"] = df[profit].diff()
df["profit_pct_change"] = df[profit].pct_change()

df["rolling_profit_mean"] = df[profit].rolling(30).mean()
df["rolling_profit_std"] = df[profit].rolling(30).std()

# ---------------------------------------------------------
# INTERACTION FEATURES
# ---------------------------------------------------------
df["price_x_volume"] = df[price] * df[volume]
df["margin_x_volume_change"] = df[mxv].diff()
df["roi_x_volume"] = df["roi"] * df[volume]

# ---------------------------------------------------------
# TIME FEATURES
# ---------------------------------------------------------
df["hour"] = df["ds"].dt.hour
df["dayofweek"] = df["ds"].dt.dayofweek
df["is_weekend"] = (df["dayofweek"] >= 5).astype(int)
df["is_peak_hours"] = df["hour"].between(18, 23).astype(int)

# ---------------------------------------------------------
# LAG FEATURES
# ---------------------------------------------------------
for lag in [1, 5, 30]:
    df[f"lag_price_{lag}"] = df[price].shift(lag)
    df[f"lag_margin_{lag}"] = df[margin].shift(lag)
    df[f"lag_volume_{lag}"] = df[volume].shift(lag)
    df[f"lag_profit_{lag}"] = df[profit].shift(lag)

# ---------------------------------------------------------
# ANOMALY FEATURES
# ---------------------------------------------------------
df["is_price_spike"] = (df[price] > df["rolling_mean_30"] + 2 * df["rolling_std_30"]).astype(int)
df["is_volume_spike"] = (df[volume] > df["rolling_volume_mean"] + 3 * df["rolling_volume_std"]).astype(int)
df["is_margin_spike"] = (df[margin] > df["rolling_margin_mean"] + 2 * df["rolling_margin_std"]).astype(int)

# ---------------------------------------------------------
# TARGET VARIABLE (predict future GP/hour)
# ---------------------------------------------------------
df["target_future_mxv"] = df[mxv].shift(-1)
cols_to_drop = ["fetched_at", "source", "item_id", "_raw_json", "timestamp"]

df = df.drop(columns=[c for c in cols_to_drop if c in df.columns])

# --- Remove timezone from ds completely ---
# Convert timezone-aware timestamps → naive timestamps
df['ds'] = pd.to_datetime(df['ds']).dt.tz_localize(None)


# ---------------------------------------------------------
# DROP NA FROM ROLLING/LAGS
# ---------------------------------------------------------
df = df.dropna()
print(df.columns.tolist())
df.head()



pd.set_option("display.max_columns", None)
df.head()
```

```python
# Correlation diagram (heatmap) for df
# Paste into a notebook cell and run. Assumes `df` is already loaded.

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# --- Settings you can change ---
corr_method = 'pearson'   # 'pearson', 'spearman', or 'kendall'
annot = True              # show correlation numbers on the heatmap
figsize = (28, 24)
cmap = "vlag"             # diverging colormap
save_dir = "ProphetScripts"
save_name = "correlation_matrix.png"

# --- Prepare data: select numeric columns and clean ---
numeric_df = df.select_dtypes(include=[np.number]).copy()
if numeric_df.shape[1] == 0:
    raise RuntimeError("No numeric columns found in df to compute correlations.")

# Replace infinities and fill NaNs (so correlation computation is stable)
numeric_df = numeric_df.replace([np.inf, -np.inf], np.nan)
numeric_df = numeric_df.fillna(method='ffill').fillna(method='bfill').fillna(0)

# Optionally drop columns with near-constant values (zero variance) to avoid misleading correlations
variance = numeric_df.var()
low_var_cols = variance[variance == 0].index.tolist()
if low_var_cols:
    numeric_df = numeric_df.drop(columns=low_var_cols)
    print(f"Dropped {len(low_var_cols)} constant columns: {low_var_cols}")

# --- Compute correlation matrix ---
corr = numeric_df.corr(method=corr_method)

# --- Plot heatmap with mask for upper triangle ---
mask = np.triu(np.ones_like(corr, dtype=bool))

plt.figure(figsize=figsize)
sns.set(style="white")
sns.heatmap(
    corr,
    mask=mask,
    cmap=cmap,
    vmax=1.0,
    vmin=-1.0,
    center=0,
    annot=annot,
    fmt=".2f",
    square=False,
    linewidths=0.5,
    cbar_kws={"shrink": 0.6, "label": f"{corr_method} correlation"}
)
plt.title(f"Correlation matrix ({corr_method}) — numeric features only", fontsize=16)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()

# --- Save figure ---
os.makedirs(save_dir, exist_ok=True)
out_path = os.path.join(save_dir, save_name)
plt.savefig(out_path, dpi=200, bbox_inches='tight')
print("Saved correlation heatmap to:", out_path)

plt.show()
```

```python
# Two-target Prophet pipeline: predict avgHighPrice and avgLowPrice (2-day forward)
# Paste into a notebook cell and run. Assumes df is already loaded and df.head() works.
# If df is not loaded, uncomment the read_csv line below.

# !pip install prophet scikit-learn matplotlib seaborn --quiet

import os
import pandas as pd
import numpy as np
from prophet import Prophet
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import TimeSeriesSplit
from sklearn.metrics import mean_absolute_error, mean_squared_error
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")

# --- Config ---
TARGETS = ['avgHighPrice', 'avgLowPrice']   # targets to model
REGRESSOR_EXCLUDE = ['ds'] + TARGETS
FORECAST_DAYS = 2
MINUTES_PER_DAY = 1440
FORECAST_HORIZON_MINUTES = FORECAST_DAYS * MINUTES_PER_DAY
N_SPLITS = 3
OUT_DIR = "ProphetScripts"
os.makedirs(OUT_DIR, exist_ok=True)

# --- 1) Basic checks and cleaning ---
print("Initial shape:", df.shape)

# Ensure ds is datetime, remove timezone, and sort
df['ds'] = pd.to_datetime(df['ds']).dt.tz_localize(None)
df = df.sort_values('ds').reset_index(drop=True)

# Verify targets exist
missing_targets = [t for t in TARGETS if t not in df.columns]
if missing_targets:
    raise ValueError(f"Missing target columns in df: {missing_targets}")

# Select numeric regressors (exclude ds and targets)
regressors = [c for c in df.columns if c not in REGRESSOR_EXCLUDE and pd.api.types.is_numeric_dtype(df[c])]
print("Using regressors:", regressors)

# Global cleaning for regressors
df[regressors] = df[regressors].replace([np.inf, -np.inf], np.nan)
for c in regressors:
    q01 = df[c].quantile(0.01)
    q99 = df[c].quantile(0.99)
    if pd.notna(q01) and pd.notna(q99) and q01 < q99:
        df[c] = df[c].clip(lower=q01, upper=q99)
df[regressors] = df[regressors].fillna(method='ffill').fillna(method='bfill').fillna(0)

# --- scaling helper ---
def scale_regressors(train_df, val_df, regressors):
    scaler = StandardScaler()
    scaler.fit(train_df[regressors])
    train_scaled = pd.DataFrame(scaler.transform(train_df[regressors]), columns=regressors, index=train_df.index)
    val_scaled = pd.DataFrame(scaler.transform(val_df[regressors]), columns=regressors, index=val_df.index)
    return train_scaled, val_scaled, scaler

# --- CV and hyperparameter grid (shared) ---
param_grid = [
    {'changepoint_prior_scale': 0.01, 'seasonality_mode': 'additive', 'seasonality_prior_scale': 10.0},
    {'changepoint_prior_scale': 0.05, 'seasonality_mode': 'additive', 'seasonality_prior_scale': 10.0},
    {'changepoint_prior_scale': 0.1,  'seasonality_mode': 'multiplicative', 'seasonality_prior_scale': 5.0},
    {'changepoint_prior_scale': 0.5,  'seasonality_mode': 'multiplicative', 'seasonality_prior_scale': 1.0},
]

def evaluate_params(params, df_tv, regressors, target_col, tscv):
    rmses = []
    maes = []
    for train_idx, val_idx in tscv.split(df_tv):
        train_split = df_tv.iloc[train_idx].copy()
        val_split = df_tv.iloc[val_idx].copy()

        # ensure ds naive
        train_split['ds'] = pd.to_datetime(train_split['ds']).dt.tz_localize(None)
        val_split['ds'] = pd.to_datetime(val_split['ds']).dt.tz_localize(None)

        # defensive cleaning inside splits
        for df_split in (train_split, val_split):
            df_split[regressors] = df_split[regressors].replace([np.inf, -np.inf], np.nan)
            for c in regressors:
                q01 = df_split[c].quantile(0.01)
                q99 = df_split[c].quantile(0.99)
                if pd.notna(q01) and pd.notna(q99) and q01 < q99:
                    df_split[c] = df_split[c].clip(lower=q01, upper=q99)
            df_split[regressors] = df_split[regressors].fillna(method='ffill').fillna(method='bfill').fillna(0)

        train_scaled, val_scaled, _ = scale_regressors(train_split, val_split, regressors)

        train_for_prophet = pd.DataFrame({'ds': train_split['ds'].values, 'y': train_split[target_col].values})
        val_for_prophet = pd.DataFrame({'ds': val_split['ds'].values, 'y': val_split[target_col].values})

        for c in regressors:
            train_for_prophet[c] = train_scaled[c].values
            val_for_prophet[c] = val_scaled[c].values

        m = Prophet(
            changepoint_prior_scale=params['changepoint_prior_scale'],
            seasonality_mode=params['seasonality_mode'],
            seasonality_prior_scale=params['seasonality_prior_scale'],
            daily_seasonality=False,
            weekly_seasonality=False,
            yearly_seasonality=False
        )
        m.add_seasonality(name='daily_minute', period=1440, fourier_order=10)
        m.add_seasonality(name='weekly_minute', period=10080, fourier_order=6)
        for r in regressors:
            m.add_regressor(r, standardize=False)

        m.fit(train_for_prophet.reset_index(drop=True))

        future = val_for_prophet[['ds'] + regressors].reset_index(drop=True).copy()
        forecast = m.predict(future)
        y_true = val_for_prophet['y'].values
        y_pred = forecast['yhat'].values

        rmses.append(np.sqrt(mean_squared_error(y_true, y_pred)))
        maes.append(mean_absolute_error(y_true, y_pred))

    return np.mean(rmses), np.mean(maes)

# --- Loop over targets: tune, fit, evaluate, forward-forecast ---
results_summary = []
tscv = TimeSeriesSplit(n_splits=N_SPLITS)

# adjust horizon if dataset too small
if len(df) <= FORECAST_HORIZON_MINUTES + 100:
    FORECAST_HORIZON_MINUTES = max(60, len(df) - 100)
    print(f"Adjusted forecast horizon to {FORECAST_HORIZON_MINUTES} minutes due to dataset size.")

test_size = FORECAST_HORIZON_MINUTES
train_val_df = df.iloc[:-test_size].copy()
test_df = df.iloc[-test_size:].copy()

# ensure ds naive in splits
train_val_df['ds'] = pd.to_datetime(train_val_df['ds']).dt.tz_localize(None)
test_df['ds'] = pd.to_datetime(test_df['ds']).dt.tz_localize(None)

for target in TARGETS:
    print("\n=== Modeling target:", target, "===\n")

    # Grid search
    results = []
    for p in param_grid:
        rmse_cv, mae_cv = evaluate_params(p, train_val_df, regressors, target, tscv)
        results.append({'params': p, 'rmse_cv': rmse_cv, 'mae_cv': mae_cv})
        print(f"Params {p} -> RMSE {rmse_cv:.4f}, MAE {mae_cv:.4f}")

    best = sorted(results, key=lambda x: x['rmse_cv'])[0]
    best_params = best['params']
    print("Best params:", best_params, "RMSE:", best['rmse_cv'])

    # Defensive cleaning on train_val_df and test_df
    for df_split in (train_val_df, test_df):
        df_split[regressors] = df_split[regressors].replace([np.inf, -np.inf], np.nan)
        for c in regressors:
            q01 = df_split[c].quantile(0.01)
            q99 = df_split[c].quantile(0.99)
            if pd.notna(q01) and pd.notna(q99) and q01 < q99:
                df_split[c] = df_split[c].clip(lower=q01, upper=q99)
        df_split[regressors] = df_split[regressors].fillna(method='ffill').fillna(method='bfill').fillna(0)

    # scale using full train_val_df scaler
    train_scaled_full, test_scaled_full, final_scaler = scale_regressors(train_val_df, test_df, regressors)

    train_prophet = pd.DataFrame({'ds': train_val_df['ds'].values, 'y': train_val_df[target].values})
    test_prophet = pd.DataFrame({'ds': test_df['ds'].values, 'y': test_df[target].values})

    for c in regressors:
        train_prophet[c] = train_scaled_full[c].values
        test_prophet[c] = test_scaled_full[c].values

    # build final model
    m_final = Prophet(
        changepoint_prior_scale=best_params['changepoint_prior_scale'],
        seasonality_mode=best_params['seasonality_mode'],
        seasonality_prior_scale=best_params['seasonality_prior_scale'],
        daily_seasonality=False,
        weekly_seasonality=False,
        yearly_seasonality=False
    )
    m_final.add_seasonality(name='daily_minute', period=1440, fourier_order=10)
    m_final.add_seasonality(name='weekly_minute', period=10080, fourier_order=6)
    for r in regressors:
        m_final.add_regressor(r, standardize=False)

    print("Fitting final model on train+val for target:", target)
    m_final.fit(train_prophet.reset_index(drop=True))

    # Forecast on holdout
    future_test = test_prophet[['ds'] + regressors].reset_index(drop=True)
    forecast_test = m_final.predict(future_test)

    y_true = test_prophet['y'].values
    y_pred = forecast_test['yhat'].values

    rmse_test = np.sqrt(mean_squared_error(y_true, y_pred))
    mae_test = mean_absolute_error(y_true, y_pred)
    mape_test = np.mean(np.abs((y_true - y_pred) / (np.where(y_true==0, 1e-8, y_true)))) * 100

    print(f"Holdout Test RMSE ({target}): {rmse_test:.4f}")
    print(f"Holdout Test MAE  ({target}): {mae_test:.4f}")
    print(f"Holdout Test MAPE ({target}): {mape_test:.2f}%")

    # Save holdout forecast
    forecast_test_out = forecast_test[['ds','yhat','yhat_lower','yhat_upper']].copy()
    forecast_test_out['y_true'] = y_true
    forecast_test_out['residual'] = forecast_test_out['y_true'] - forecast_test_out['yhat']
    out_path_hold = os.path.join(OUT_DIR, f"forecast_holdout_{target}.csv")
    forecast_test_out.to_csv(out_path_hold, index=False)
    print("Saved holdout forecast to:", out_path_hold)

    # --- Forward forecast beyond last timestamp in df (2-day forward) ---
    N_minutes_forward = FORECAST_DAYS * MINUTES_PER_DAY
    last_row = df.iloc[-1]
    future_rows = []
    for i in range(1, N_minutes_forward + 1):
        new_ds = last_row['ds'] + pd.Timedelta(minutes=i)
        row = {'ds': new_ds}
        for r in regressors:
            row[r] = last_row[r]  # carry-forward baseline for regressors
        future_rows.append(row)
    future_df = pd.DataFrame(future_rows)

    # scale using final_scaler
    future_scaled = pd.DataFrame(final_scaler.transform(future_df[regressors]), columns=regressors)
    future_for_prophet = pd.concat([future_df[['ds']].reset_index(drop=True), future_scaled.reset_index(drop=True)], axis=1)
    future_forecast = m_final.predict(future_for_prophet)

    forward_out = future_forecast[['ds','yhat','yhat_lower','yhat_upper']].copy()
    out_path_forward = os.path.join(OUT_DIR, f"forecast_forward_2day_{target}.csv")
    forward_out.to_csv(out_path_forward, index=False)
    print("Saved 2-day forward forecast to:", out_path_forward)

    # Plot forward forecast sample
    plt.figure(figsize=(12,4))
    plt.plot(forward_out['ds'], forward_out['yhat'], label=f'Predicted {target}')
    plt.fill_between(forward_out['ds'], forward_out['yhat_lower'], forward_out['yhat_upper'], color='gray', alpha=0.2)
    plt.title(f"2-day forward forecast for {target}")
    plt.xlabel("ds")
    plt.ylabel(target)
    plt.legend()
    plt.tight_layout()
    plt.show()

    # Save model and scaler references to disk (optional)
    # If you want to reuse the model/scaler in the notebook, store them in a dict
    results_summary.append({
        'target': target,
        'best_params': best_params,
        'rmse_test': rmse_test,
        'mae_test': mae_test,
        'mape_test': mape_test,
        'holdout_csv': out_path_hold,
        'forward_csv': out_path_forward
    })

# --- Summary ---
summary_df = pd.DataFrame(results_summary)
summary_path = os.path.join(OUT_DIR, "modeling_summary_avgHigh_avgLow.csv")
summary_df.to_csv(summary_path, index=False)
print("\nModeling summary saved to:", summary_path)
display(summary_df)
```

```python
# Fixed: clear buy / sell recommendation printout
# Requirements: run after you have produced a forward forecast DataFrame named `future_forecast`
# and your original dataframe `df` is available. This cell will:
#  - recreate a simple signal if signals_df is missing
#  - print a single clear recommendation line like:
#       BUY at X.XX
#       SELL for Y.YY
#  - be robust to missing columns and small edge cases

import pandas as pd
import numpy as np

def fmt_dt(x):
    if pd.isna(x) or x is None:
        return "N/A"
    try:
        return pd.to_datetime(x).strftime("%Y-%m-%d %H:%M")
    except Exception:
        return str(x)

# Ensure we have either signals_df or future_forecast
if 'signals_df' in globals() and isinstance(signals_df, pd.DataFrame) and len(signals_df) > 0:
    s = signals_df.iloc[0].to_dict()
else:
    # Recreate a minimal signal from future_forecast
    if 'future_forecast' not in globals():
        raise RuntimeError("No signals_df or future_forecast found. Run the model and forward-forecast cells first.")
    ff = future_forecast.copy().reset_index(drop=True)

    # Determine baseline price: prefer avgHighPrice, then avgLowPrice, then yhat
    last_price = np.nan
    for col in ['avgHighPrice', 'avgLowPrice', 'price_x_volume', 'price_change']:
        if col in globals().get('df', pd.DataFrame()).columns:
            last_price = df.iloc[-1][col]
            break
    if pd.isna(last_price):
        # fallback to first forecasted yhat
        if 'yhat' in ff.columns and len(ff) > 0:
            last_price = float(ff['yhat'].iloc[0])
        else:
            raise RuntimeError("Could not determine a baseline price from df or future_forecast.")

    # Strategy parameters (tweakable)
    profit_target_pct = 0.02   # target profit (2%)
    stop_loss_pct = 0.01       # stop loss (1%)
    slippage_pct = 0.001       # assumed slippage on entry

    entry_price = float(last_price) * (1 + slippage_pct)
    target_price = entry_price * (1 + profit_target_pct)
    stop_price = entry_price * (1 - stop_loss_pct)

    # Ensure forecast columns exist
    if 'yhat' not in ff.columns:
        raise RuntimeError("future_forecast must contain a 'yhat' column.")
    ff['pred'] = ff['yhat'].astype(float)
    # use yhat_lower / yhat_upper if present for confidence calc
    ff['pred_low'] = ff['yhat_lower'] if 'yhat_lower' in ff.columns else ff['pred']
    ff['pred_high'] = ff['yhat_upper'] if 'yhat_upper' in ff.columns else ff['pred']

    # Find earliest crossing of target or stop
    hit_target = ff[ff['pred'] >= target_price]
    hit_stop = ff[ff['pred'] <= stop_price]

    first_target = hit_target.iloc[0] if len(hit_target) > 0 else None
    first_stop = hit_stop.iloc[0] if len(hit_stop) > 0 else None

    # Build signal dict `s`
    if first_target is not None and (first_stop is None or pd.to_datetime(first_target['ds']) <= pd.to_datetime(first_stop['ds'])):
        s = {
            'signal': 'MODEL_BUY_SIGNAL',
            'entry_price_estimate': entry_price,
            'target_price': float(first_target['pred']),
            'expected_exit_time': first_target['ds'],
            'expected_exit_price': float(first_target['pred'])
        }
    elif first_stop is not None:
        s = {
            'signal': 'MODEL_SELL_SIGNAL',
            'entry_price_estimate': entry_price,
            'stop_price': float(first_stop['pred']),
            'expected_exit_time': first_stop['ds'],
            'expected_exit_price': float(first_stop['pred'])
        }
    else:
        s = {
            'signal': 'HOLD',
            'entry_price_estimate': entry_price,
            'target_price': target_price,
            'stop_price': stop_price
        }

# Print a concise human-readable recommendation
signal = s.get('signal', 'HOLD')
entry_price = s.get('entry_price_estimate', None)
target_price = s.get('target_price', None)
stop_price = s.get('stop_price', None)
exit_time = s.get('expected_exit_time', None)
exit_price = s.get('expected_exit_price', None)

if signal == 'MODEL_BUY_SIGNAL':
    # Print buy and sell lines
    print(f"BUY at {entry_price:.2f}")
    if target_price is not None:
        print(f"SELL for {float(target_price):.2f}")
    if exit_time is not None:
        print(f"Expected sell time: {fmt_dt(exit_time)}")
elif signal == 'MODEL_SELL_SIGNAL':
    # Model expects decline — show actionable lines for holders
    print("DO NOT BUY — model expects a decline")
    if entry_price is not None:
        print(f"Current baseline price: {entry_price:.2f}")
    if stop_price is not None:
        print(f"Expected drop to: {float(stop_price):.2f}")
    if exit_time is not None:
        print(f"Expected time of drop: {fmt_dt(exit_time)}")
else:
    # HOLD
    print("HOLD — no clear buy/sell signal within the forecast horizon")
    if entry_price is not None:
        print(f"Baseline price: {entry_price:.2f}")
    if target_price is not None and stop_price is not None:
        print(f"Example target: {float(target_price):.2f}  stop: {float(stop_price):.2f}")
```
