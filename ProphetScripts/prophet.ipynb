```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

root = Path.cwd()
data_dir = root / "data"
if not data_dir.exists():
    data_dir = root.parent / "data"



csv_files = {
    "chaos_rune_562_1year_daily": data_dir / "chaos_rune_562_1year_daily.csv",
    "chaos_rune_12929_lastweek_minute": data_dir / "chaos_rune_12929_lastweek_minute.csv",
    "item_562_last_24_hours": data_dir / "item_562_last_24_hours.csv",
}

frames = {}
for name, path in csv_files.items():
    frames[name] = pd.read_csv(path)

frames
```

```python
import pandas as pd
import numpy as np

# ---------------------------------------------------------
# LOAD DATA CORRECTLY
# ---------------------------------------------------------
df = frames["chaos_rune_12929_lastweek_minute"].copy()

# ds is already ISO8601 with timezone → convert safely
df["ds"] = pd.to_datetime(df["ds"], errors="coerce")

# Basic cleanup
df = df.sort_values("ds")
df = df.reset_index(drop=True)

# ---------------------------------------------------------
# COLUMN SHORTCUTS
# ---------------------------------------------------------
price = "avgHighPrice"
sell = "avgLowPrice"
margin = "margin"
volume = "daily_volume"
profit = "potential_profit"
mxv = "margin_x_volume"

# ---------------------------------------------------------
# PRICE FEATURES
# ---------------------------------------------------------
df["price_change"] = df[price].diff()
df["price_pct_change"] = df[price].pct_change()

df["rolling_mean_5"] = df[price].rolling(5).mean()
df["rolling_mean_30"] = df[price].rolling(30).mean()
df["rolling_std_30"] = df[price].rolling(30).std()

df["zscore_price"] = (df[price] - df["rolling_mean_30"]) / df["rolling_std_30"]

# ---------------------------------------------------------
# MARGIN FEATURES
# ---------------------------------------------------------
df["margin_change"] = df[margin].diff()
df["margin_pct_change"] = df[margin].pct_change()

df["rolling_margin_mean"] = df[margin].rolling(30).mean()
df["rolling_margin_std"] = df[margin].rolling(30).std()

# ---------------------------------------------------------
# VOLUME FEATURES
# ---------------------------------------------------------
df["volume_change"] = df[volume].diff()
df["volume_pct_change"] = df[volume].pct_change()

df["rolling_volume_mean"] = df[volume].rolling(30).mean()
df["rolling_volume_std"] = df[volume].rolling(30).std()

# ---------------------------------------------------------
# PROFIT FEATURES
# ---------------------------------------------------------
df["profit_change"] = df[profit].diff()
df["profit_pct_change"] = df[profit].pct_change()

df["rolling_profit_mean"] = df[profit].rolling(30).mean()
df["rolling_profit_std"] = df[profit].rolling(30).std()

# ---------------------------------------------------------
# INTERACTION FEATURES
# ---------------------------------------------------------
df["price_x_volume"] = df[price] * df[volume]
df["margin_x_volume_change"] = df[mxv].diff()
df["roi_x_volume"] = df["roi"] * df[volume]

# ---------------------------------------------------------
# TIME FEATURES
# ---------------------------------------------------------
df["hour"] = df["ds"].dt.hour
df["dayofweek"] = df["ds"].dt.dayofweek
df["is_weekend"] = (df["dayofweek"] >= 5).astype(int)
df["is_peak_hours"] = df["hour"].between(18, 23).astype(int)

# ---------------------------------------------------------
# LAG FEATURES
# ---------------------------------------------------------
for lag in [1, 5, 30]:
    df[f"lag_price_{lag}"] = df[price].shift(lag)
    df[f"lag_margin_{lag}"] = df[margin].shift(lag)
    df[f"lag_volume_{lag}"] = df[volume].shift(lag)
    df[f"lag_profit_{lag}"] = df[profit].shift(lag)

# ---------------------------------------------------------
# ANOMALY FEATURES
# ---------------------------------------------------------
df["is_price_spike"] = (df[price] > df["rolling_mean_30"] + 2 * df["rolling_std_30"]).astype(int)
df["is_volume_spike"] = (df[volume] > df["rolling_volume_mean"] + 3 * df["rolling_volume_std"]).astype(int)
df["is_margin_spike"] = (df[margin] > df["rolling_margin_mean"] + 2 * df["rolling_margin_std"]).astype(int)

# ---------------------------------------------------------
# TARGET VARIABLE (predict future GP/hour)
# ---------------------------------------------------------
df["target_future_mxv"] = df[mxv].shift(-1)
cols_to_drop = ["fetched_at", "source", "item_id", "_raw_json", "timestamp"]

df = df.drop(columns=[c for c in cols_to_drop if c in df.columns])

# --- Remove timezone from ds completely ---
# Convert timezone-aware timestamps → naive timestamps
df['ds'] = pd.to_datetime(df['ds']).dt.tz_localize(None)


# ---------------------------------------------------------
# DROP NA FROM ROLLING/LAGS
# ---------------------------------------------------------
df = df.dropna()
pd.set_option("display.max_columns", None)
print(df.columns.tolist())
df.head()
```

```python
# Correlation diagram (heatmap) for df
# Paste into a notebook cell and run. Assumes `df` is already loaded.

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# --- Settings you can change ---
corr_method = 'pearson'   # 'pearson', 'spearman', or 'kendall'
annot = True              # show correlation numbers on the heatmap
figsize = (32, 18)
cmap = "vlag"             # diverging colormap
save_dir = "ProphetScripts"
save_name = "correlation_matrix.png"

# --- Prepare data: select numeric columns and clean ---
numeric_df = df.select_dtypes(include=[np.number]).copy()
if numeric_df.shape[1] == 0:
    raise RuntimeError("No numeric columns found in df to compute correlations.")

# Replace infinities and fill NaNs (so correlation computation is stable)
numeric_df = numeric_df.replace([np.inf, -np.inf], np.nan)
numeric_df = numeric_df.fillna(method='ffill').fillna(method='bfill').fillna(0)

# Optionally drop columns with near-constant values (zero variance) to avoid misleading correlations
variance = numeric_df.var()
low_var_cols = variance[variance == 0].index.tolist()
if low_var_cols:
    numeric_df = numeric_df.drop(columns=low_var_cols)
    print(f"Dropped {len(low_var_cols)} constant columns: {low_var_cols}")

# --- Compute correlation matrix ---
corr = numeric_df.corr(method=corr_method)

# --- Plot heatmap with mask for upper triangle ---
mask = np.triu(np.ones_like(corr, dtype=bool))

plt.figure(figsize=figsize)
sns.set(style="white")
sns.heatmap(
    corr,
    mask=mask,
    cmap=cmap,
    vmax=1.0,
    vmin=-1.0,
    center=0,
    annot=annot,
    fmt=".2f",
    square=False,
    linewidths=0.5,
    cbar_kws={"shrink": 0.6, "label": f"{corr_method} correlation"}
)
plt.title(f"Correlation matrix ({corr_method}) — numeric features only", fontsize=16)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()

# --- Save figure ---
os.makedirs(save_dir, exist_ok=True)
out_path = os.path.join(save_dir, save_name)
plt.savefig(out_path, dpi=200, bbox_inches='tight')
print("Saved correlation heatmap to:", out_path)

plt.show()
```

```python
# -----------------------------
# CONFIG + MODES
# -----------------------------

MODE = 3   # <-- change this between 1, 2, 3

# Mode definitions
MODES = {
    1: {  # Next 24 hours (minute + hour)
        "forecast_horizon_days": 1,
        "forecast_freq": "min",
        "plot_freq": "h"
    },
    2: {  # Next 2 days (minute + hour)
        "forecast_horizon_days": 2,
        "forecast_freq": "min",
        "plot_freq": "h"
    },
    3: {  # Next 7 days (hour only)
        "forecast_horizon_days": 7,
        "forecast_freq": "h",
        "plot_freq": "h"
    },
    4: {  # Seasonality-focused mode
    "forecast_horizon_days": 7,
    "forecast_freq": "h",
    "plot_freq": "h",

    # Prophet overrides
    "growth": "linear",
    "seasonality_mode": "multiplicative",
    "changepoint_prior_scale": 0.05,   # very low = trend is stable
    "changepoint_range": 0.3,
    "interval_width": 0.35,
    "uncertainty_samples": 80,

    # Seasonality emphasis
    "daily_seasonality": True,
    "weekly_seasonality": True,
    "fourier_daily": 10,               # stronger daily pattern
    "fourier_weekly": 15,              # stronger weekly pattern
    "custom_seasonality_period": 7,
    "custom_seasonality_fourier": 8    # stronger OSRS weekend cycle
}


}
```

```python
# Apply selected mode
FORECAST_HORIZON_DAYS = MODES[MODE]["forecast_horizon_days"]
FORECAST_FREQ = MODES[MODE]["forecast_freq"]
PLOT_RESAMPLE_FREQ = MODES[MODE]["plot_freq"]

# -----------------------------
# CONFIG
# -----------------------------
RESAMPLE_FREQ = "min"
CUSTOM_SEASONALITY_NAME = "osrs_weekend"
SMOOTH_LOW_W = 3
SMOOTH_HIGH_W = 6
OUTLIER_LOOKBACK_HOURS = 24 * 30
REGRESSORS = [
    "highPriceVolume", "lowPriceVolume", "daily_volume", "roi", "margin",
    "potential_profit", "margin_x_volume", "price_change"
]
if MODE == 4:
    PROPHET_GROWTH = MODES[4]["growth"]
    PROPHET_SEASONALITY_MODE = MODES[4]["seasonality_mode"]
    PROPHET_CHANGPOINT_PRIOR_SCALE = MODES[4]["changepoint_prior_scale"]
    PROPHET_CHANGPOINT_RANGE = MODES[4]["changepoint_range"]
    PROPHET_INTERVAL_WIDTH = MODES[4]["interval_width"]
    PROPHET_UNCERTAINTY_SAMPLES = MODES[4]["uncertainty_samples"]

    PROPHET_DAILY_SEASONALITY = MODES[4]["daily_seasonality"]
    PROPHET_WEEKLY_SEASONALITY = MODES[4]["weekly_seasonality"]
    CUSTOM_SEASONALITY_PERIOD = MODES[4]["custom_seasonality_period"]
    CUSTOM_SEASONALITY_FOURIER = MODES[4]["custom_seasonality_fourier"]
else:   
    PROPHET_GROWTH = "linear"
    PROPHET_SEASONALITY_MODE = "additive"
    PROPHET_CHANGPOINT_PRIOR_SCALE = 0.15
    PROPHET_CHANGPOINT_RANGE = 0.4
    PROPHET_INTERVAL_WIDTH = 0.4
    PROPHET_UNCERTAINTY_SAMPLES = 50

    PROPHET_DAILY_SEASONALITY = True
    PROPHET_WEEKLY_SEASONALITY = True
    CUSTOM_SEASONALITY_PERIOD = 7
    CUSTOM_SEASONALITY_FOURIER = 5

FUTURE_REGRESSOR_ROLLING_WINDOW = 60 * 24
PLOT_FIGSIZE = (16, 9)

config = {
    "resample_freq": RESAMPLE_FREQ,
    "smoothing_windows": {"avgLowPrice": SMOOTH_LOW_W, "avgHighPrice": SMOOTH_HIGH_W},
    "regressors": REGRESSORS,
    "outlier_lookback_hours": OUTLIER_LOOKBACK_HOURS,
    "prophet": {
        "growth": PROPHET_GROWTH,
        "daily_seasonality": PROPHET_DAILY_SEASONALITY,
        "weekly_seasonality": PROPHET_WEEKLY_SEASONALITY,
        "seasonality_mode": PROPHET_SEASONALITY_MODE,
        "changepoint_prior_scale": PROPHET_CHANGPOINT_PRIOR_SCALE,
        "changepoint_range": PROPHET_CHANGPOINT_RANGE,
        "interval_width": PROPHET_INTERVAL_WIDTH,
        "uncertainty_samples": PROPHET_UNCERTAINTY_SAMPLES,
    },
    "custom_seasonality": {
        "name": CUSTOM_SEASONALITY_NAME,
        "period": CUSTOM_SEASONALITY_PERIOD,
        "fourier_order": CUSTOM_SEASONALITY_FOURIER,
    },
    "forecast_horizon_days": FORECAST_HORIZON_DAYS,
    "forecast_freq": FORECAST_FREQ,
    "future_regressor_rolling_window": FUTURE_REGRESSOR_ROLLING_WINDOW,
    "plot_figsize": PLOT_FIGSIZE,
    "plot_resample_freq": PLOT_RESAMPLE_FREQ
}
```

```python
def detect_outliers(series: pd.Series) -> pd.Series:
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return (series < lower) | (series > upper)


def build_prophet_model(cfg: dict) -> Prophet:
    p = cfg["prophet"]
    m = Prophet(
        growth=p["growth"],
        daily_seasonality=p["daily_seasonality"],
        weekly_seasonality=p["weekly_seasonality"],
        seasonality_mode=p["seasonality_mode"],
        changepoint_prior_scale=p["changepoint_prior_scale"],
        changepoint_range=p["changepoint_range"],
        interval_width=p["interval_width"],
        uncertainty_samples=p["uncertainty_samples"],
    )

    cs = cfg["custom_seasonality"]
    m.add_seasonality(
        name=cs["name"],
        period=cs["period"],
        fourier_order=cs["fourier_order"],
    )

    for r in cfg["regressors"]:
        m.add_regressor(r)

    return m


def fill_regressors(df: pd.DataFrame, regressors: list[str]) -> pd.DataFrame:
    df[regressors] = df[regressors].ffill().bfill()
    return df


def compute_future_regressors(df_low: pd.DataFrame, df_high: pd.DataFrame,
                              regressors: list[str], rolling_window: int) -> tuple[dict, dict]:
    future_low_vals = {}
    future_high_vals = {}

    for r in regressors:
        series_low = df_low[r]
        series_high = df_high[r]

        lv_low = series_low.rolling(rolling_window, min_periods=1).mean().iloc[-1]
        lv_high = series_high.rolling(rolling_window, min_periods=1).mean().iloc[-1]

        if pd.isna(lv_low):
            lv_low = series_low.iloc[-1]
        if pd.isna(lv_high):
            lv_high = series_high.iloc[-1]

        future_low_vals[r] = lv_low
        future_high_vals[r] = lv_high

    return future_low_vals, future_high_vals
```

```python
# -----------------------------
# 1. PREPARE DATA
# -----------------------------
df["ds"] = pd.to_datetime(df["ds"])
df_resampled = df.resample(config["resample_freq"], on="ds").mean().reset_index()

# -----------------------------
# 2. SMOOTH PRICE SIGNALS
# -----------------------------
df_resampled["avgLowPrice"] = (
    df_resampled["avgLowPrice"]
    .rolling(window=config["smoothing_windows"]["avgLowPrice"], min_periods=1)
    .mean()
)
df_resampled["avgHighPrice"] = (
    df_resampled["avgHighPrice"]
    .rolling(window=config["smoothing_windows"]["avgHighPrice"], min_periods=1)
    .mean()
)

# -----------------------------
# 3. CLEAN REGRESSORS
# -----------------------------
regressors = config["regressors"]
df_resampled = fill_regressors(df_resampled, regressors)
df_resampled["avgLowPrice"] = df_resampled["avgLowPrice"].ffill().bfill()
df_resampled["avgHighPrice"] = df_resampled["avgHighPrice"].ffill().bfill()

# -----------------------------
# 4. OUTLIER DETECTION
# -----------------------------
lookback_periods = int(config["outlier_lookback_hours"])
last_window = df_resampled.tail(lookback_periods)

df_resampled["low_outlier"] = detect_outliers(last_window["avgLowPrice"]).reindex(
    df_resampled.index, fill_value=False
)
df_resampled["high_outlier"] = detect_outliers(last_window["avgHighPrice"]).reindex(
    df_resampled.index, fill_value=False
)

low_outliers = df_resampled[df_resampled["low_outlier"]]
high_outliers = df_resampled[df_resampled["high_outlier"]]

# -----------------------------
# 5. PREPARE PROPHET DATAFRAMES
# -----------------------------
df_low = (
    df_resampled.copy()
    .rename(columns={"avgLowPrice": "y"})
    .dropna(subset=["y"])
)
df_high = (
    df_resampled.copy()
    .rename(columns={"avgHighPrice": "y"})
    .dropna(subset=["y"])
)

df_low = fill_regressors(df_low, regressors)
df_high = fill_regressors(df_high, regressors)

df_low["floor"] = df_low["y"].min() * 0.95
df_low["cap"] = df_low["y"].max() * 1.05
df_high["floor"] = df_high["y"].min() * 0.95
df_high["cap"] = df_high["y"].max() * 1.05
```

```python
# -----------------------------
# 6. BUILD MODELS
# -----------------------------
m_low = build_prophet_model(config)
m_high = build_prophet_model(config)

# -----------------------------
# 7. FIT MODELS
# -----------------------------
fit_cols_low = ["ds", "y", "floor"] + regressors
fit_cols_high = ["ds", "y"] + regressors

m_low.fit(df_low[fit_cols_low])
m_high.fit(df_high[fit_cols_high])

# -----------------------------
# 8. FORECAST NEXT N DAYS
# -----------------------------
# Horizon calculation based on mode frequency
if config["forecast_freq"] == "min":
    periods = config["forecast_horizon_days"] * 24 * 60   # minutes
elif config["forecast_freq"] == "h":
    periods = config["forecast_horizon_days"] * 24        # hours
else:
    raise ValueError("Unsupported forecast frequency")

future_low = m_low.make_future_dataframe(periods=int(periods), freq=config["forecast_freq"])
future_high = m_high.make_future_dataframe(periods=int(periods), freq=config["forecast_freq"])


future_low["floor"] = df_low["floor"].iloc[-1]

rolling_window = config["future_regressor_rolling_window"]
future_low_vals, future_high_vals = compute_future_regressors(
    df_low, df_high, regressors, rolling_window
)

for r in regressors:
    future_low[r] = future_low_vals[r]
    future_high[r] = future_high_vals[r]

future_low = fill_regressors(future_low, regressors)
future_high = fill_regressors(future_high, regressors)

forecast_low = m_low.predict(future_low)
forecast_high = m_high.predict(future_high)

# -----------------------------
# 9. SOFT CLIPPING
# -----------------------------

min_low = df_low["y"].min() * 0.9
max_low = df_low["y"].max() * 1.1
min_high = df_high["y"].min() * 0.9
max_high = df_high["y"].max() * 1.1

forecast_low["yhat"] = forecast_low["yhat"].clip(lower=min_low, upper=max_low)
forecast_high["yhat"] = forecast_high["yhat"].clip(lower=min_high, upper=max_high)

```

```python
# -----------------------------
# 10. RESAMPLE FOR PLOTTING
# -----------------------------
forecast_low_hourly = forecast_low.resample("h", on="ds").mean().reset_index()
forecast_high_hourly = forecast_high.resample("h", on="ds").mean().reset_index()

# -----------------------------
# 11. PLOT
# -----------------------------
plt.figure(figsize=config["plot_figsize"])

plt.plot(
    df_resampled["ds"],
    df_resampled["avgLowPrice"],
    color="#1f77b4",
    label="Actual Low Price",
)
plt.plot(
    df_resampled["ds"],
    df_resampled["avgHighPrice"],
    color="#d62728",
    label="Actual High Price",
)

plt.plot(
    forecast_low_hourly["ds"],
    forecast_low_hourly["yhat"],
    color="#1f77b4",
    linestyle="--",
    label="Forecast Low Price",
)
plt.plot(
    forecast_high_hourly["ds"],
    forecast_high_hourly["yhat"],
    color="#d62728",
    linestyle="--",
    label="Forecast High Price",
)

plt.fill_between(
    forecast_low_hourly["ds"],
    forecast_low_hourly["yhat_lower"],
    forecast_low_hourly["yhat_upper"],
    color="#1f77b4",
    alpha=0.15,
    label="Low Price Uncertainty",
)
plt.fill_between(
    forecast_high_hourly["ds"],
    forecast_high_hourly["yhat_lower"],
    forecast_high_hourly["yhat_upper"],
    color="#d62728",
    alpha=0.15,
    label="High Price Uncertainty",
)

plt.title("OSRS Item Forecast", fontsize=18, fontweight="bold")
plt.xlabel("Date", fontsize=14)
plt.ylabel("Price", fontsize=14)
plt.grid(True, alpha=0.25)
plt.legend(fontsize=12)
plt.tight_layout()
plt.show()

# -----------------------------
# 12. PRINT SUMMARY
# -----------------------------
print("\nLow Price Forecast (hourly):")
print(forecast_low_hourly[["ds", "yhat"]].tail())

print("\nHigh Price Forecast (hourly):")
print(forecast_high_hourly[["ds", "yhat"]].tail())

print("\nDetected Low Price Outliers:")
print(low_outliers[["ds", "avgLowPrice"]].tail(10))

print("\nDetected High Price Outliers:")
print(high_outliers[["ds", "avgHighPrice"]].tail(10))
# -----------------------------
# 13. TREND COMPONENT PLOTS
# -----------------------------
from prophet.plot import plot_components

# Plot trend and seasonality for low price model
fig_low = plot_components(m_low, forecast_low)
fig_low.suptitle("Low Price Model Components", fontsize=16, fontweight="bold")

# Plot trend and seasonality for high price model
fig_high = plot_components(m_high, forecast_high)
fig_high.suptitle("High Price Model Components", fontsize=16, fontweight="bold")

plt.show()
```

```python
# -----------------------------
# 14. MODEL-BASED BUY/SELL WINDOWS (NEXT 24H & 7D, SAFE OSRS LOGIC)
# -----------------------------
import numpy as np
import pandas as pd

# Merge hourly forecasts
low_fc = forecast_low_hourly[['ds', 'yhat', 'yhat_lower', 'yhat_upper']].rename(
    columns={'yhat': 'low_yhat', 'yhat_lower': 'low_yhat_lower', 'yhat_upper': 'low_yhat_upper'}
)
high_fc = forecast_high_hourly[['ds', 'yhat', 'yhat_lower', 'yhat_upper']].rename(
    columns={'yhat': 'high_yhat', 'yhat_lower': 'high_yhat_lower', 'yhat_upper': 'high_yhat_upper'}
)
fc = low_fc.merge(high_fc, on='ds', how='inner')

# Current time
now = pd.Timestamp.now()

# Filter forecast windows
fc_24h = fc[(fc['ds'] >= now) & (fc['ds'] <= now + pd.Timedelta(hours=24))]
fc_7d  = fc[(fc['ds'] >= now) & (fc['ds'] <= now + pd.Timedelta(days=7))]

def summarize_window(window_df, label: str):
    print(f"\n=== {label} MODEL-BASED OSRS PROFIT WINDOW (NOT FINANCIAL ADVICE) ===")

    if window_df.empty:
        print("No forecast data available.")
        return

    # BUY = lowest predicted low price
    buy_row = window_df.loc[window_df['low_yhat'].idxmin()]

    # SELL = highest predicted high price
    sell_row = window_df.loc[window_df['high_yhat'].idxmax()]

    buy_time = buy_row['ds']
    sell_time = sell_row['ds']

    buy_price = float(buy_row['low_yhat'])
    sell_price = float(sell_row['high_yhat'])
    profit = sell_price - buy_price

    # Reject impossible trades (sell before buy)
    if sell_time <= buy_time:
        print("No profitable trade detected: SELL happens before BUY.")
        print(f"Suggested BUY time:  {buy_time}")
        print(f"Suggested SELL time: {sell_time}")
        return

    # Reject negative or zero profit
    if profit <= 0:
        print("No profitable trade detected: profit is zero or negative.")
        print(f"BUY price:  {buy_price:.2f} gp")
        print(f"SELL price: {sell_price:.2f} gp")
        return

    print("Suggested BUY:")
    print(f"  Time:  {buy_time}")
    print(f"  Price: {buy_price:.2f} gp")

    print("Suggested SELL:")
    print(f"  Time:  {sell_time}")
    print(f"  Price: {sell_price:.2f} gp")

    print(f"\nEstimated Profit: {profit:.2f} gp per item")

# 24-hour profit window (from now)
summarize_window(fc_24h, "NEXT 24H")

# 7-day profit window (from now)
summarize_window(fc_7d, "NEXT 7D")
```
